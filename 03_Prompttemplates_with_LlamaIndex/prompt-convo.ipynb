```python
%pip install llama-index-llms-openai
from llama_index.core.llms import ChatMessage
from llama_index.llms.openai import OpenAI
```

```python
messages = [
        ChatMessage(role="system", content="You are a helpful assistant that translates English to French. You will be given an English sentence, and you will respond with the French translation."),
        ChatMessage(role="user", content="Translate the following English sentence to French: 'Hello, how are you?'")
]
resp =OpenAI().chat(messages)
print(resp)
```

```python
messages =[
    ChatMessage(role="system",content="I am an HR strategist of a large corporation.I am about to start a meeting with the CEO and CFO to discuss the company's hiring strategy for the next year. I want you to act as my assistant and provide me with insights and recommendations based on the latest trends in the job market, as well as the company's current hiring needs and goals." \
    "I Need to prepare a presentation for the meeting.  I need to present the latest trends in hiring and workforce development."),
    ChatMessage(role="user",content="What are the latest trends in hiring and workforce development that I should include in my presentation?")
    ]
resp = OpenAI().chat(messages)
print(resp)
```
