A ReAct (Reason + Act) agent is an AI framework that combines Large Language Model (LLM) reasoning with external tools to solve complex tasks, interacting with environments through a "Thought, Action, Observation" loop. It moves beyond static text generation to perform multi-step planning, such as searching the web, querying databases, or using APIs, allowing it to act, observe, and refine its strategy

This project is about building a Calculator,tools are defined for each calculation function.Prompts provide a natural language question.React agent runs stepwise to generate the right outcome by using right tools.

```python
from llama_index.core.agent import ReActAgent
from llama_index.llms.openai import OpenAI
from llama_index.core.tools import FunctionTool
```

```python
def add(a:float,b:float) -> float:
    """Add two numbers."""
    return a + b
add_tool = FunctionTool.from_defaults(fn=add)

def multiply(a:float,b:float) -> float:
    """Multiply two numbers."""
    return a * b
multiply_tool = FunctionTool.from_defaults(fn=multiply)

def substract(a:float,b:float) -> float:
    """Subtract two numbers."""
    return a - b
substract_tool = FunctionTool.from_defaults(fn=substract)

def divide(a:float,b:float) -> float:
    """Divide two numbers."""
    if b==0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b
divide_tool = FunctionTool.from_defaults(fn=divide)

#scientific Functions
def power(a:float,b:float) -> float:
    """Calculate a to the power of b."""
    return a ** b   
power_tool = FunctionTool.from_defaults(fn=power)

def factorial(n: int) -> int:
    """Calculate the factorial of n."""
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    elif n == 0 or n == 1:
        return 1
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result
factorial_tool = FunctionTool.from_defaults(fn=factorial)

```

```python
tools =[multiply_tool,add_tool,substract_tool,divide_tool,power_tool,factorial_tool]
```

```python
llm = OpenAI(model="gpt-3.5-turbo")
agent =ReActAgent.from_tools(tools,llm=llm,verbose=True)
```

```python
response = agent.chat("What is 20+(2*4) ? Calculate step by step")
```
